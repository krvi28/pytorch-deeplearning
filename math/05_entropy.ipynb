#### Entropy

```python
# Load libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# probability of an event happening
p = 0.25

# Not the correct formula
H = -(p*np.log(p))
print('Wrong Entropy: ', str(H))
```

```python
# probability of an event happening
p = 0.25
q = 1 - p # probability of the event not happening

x = [0.25, 0.75]

H = 0
for p in x:
  H += -(p*np.log(p))

print('Correct Entropy: ', str(H))
```

```python
# Another way, written for n=2 events
H = -((p*np.log(p)) + ((1-p)*np.log(1-p)))
print('Correct Entropy: ', str(H))
```

#### Cross-entropy

```python
# Note: all probabilities must sum to 1
p = [1, 0]
q = [0.25, 0.75]

H = 0
for i in range(len(p)):
  H += -(p[i]*np.log(q[i]))
print('Cross-entroy: ', H)
```

```python
# Using pyTorch
import torch
import torch.nn.functional as F
```

```python
# note: inputs must be tensors
p_tensor = torch.Tensor(p)
q_tensor = torch.Tensor(q)

#F.binary_cross_entropy(p, q)
#F.binary_cross_entropy(p_tensor, q_tensor)
F.binary_cross_entropy(q_tensor, p_tensor) # binary cross entropy function is sensitive to the order of the inputs
```
