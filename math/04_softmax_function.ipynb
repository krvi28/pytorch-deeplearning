```python
# Load libraries
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
# FUnction to implement softmax

def softmax(data: list) -> float:
  numerator = np.exp(data)
  denominator = np.sum(numerator)
  return numerator/denominator
```

```python
# Implement Sigmoid function using numpy manually

z = [1, 2, 3, 4, 5]

result = softmax(z)

print(result)
print(result.sum())
```

```python
z = np.random.randint(-5, high = 15, size = 35)

result = softmax(z)

print(result)
print(result.sum())

plt.scatter(z, result)
plt.xlabel("Input")
plt.ylabel(r"Softmax - $\sigma$")
# plt.yscale('log')
# In log scale, the input values are linearly transformed into output values that are scaled between 0 and 1
# and sum to 1
plt.title("Softmax Function");
```

```python
# In PyTorch

# create an instance of the softmax activation function class
softmax_cls = torch.nn.Softmax(dim=0)

# apply the function to the data
z = [1, 2, 3, 4, 5]
sigma = softmax_cls(torch.Tensor(z))

print(sigma)
print(sigma.sum())

plt.scatter(z, sigma)
plt.xlabel("Input")
plt.ylabel(r"Softmax - $\sigma$")
# plt.yscale('log')
# In log scale, the input values are linearly transformed into output values that are scaled between 0 and 1
# and sum to 1
plt.title("Softmax Function - PyTorch");
```
